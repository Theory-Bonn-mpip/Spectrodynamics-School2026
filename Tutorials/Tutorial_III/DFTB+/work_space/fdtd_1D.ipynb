# FDTD 1D Tutorial

The objective of this tutorial is to learn the basics of running MARADO and the steps required to simulate the electronic dynamics of a molecule in the electronic strong-coupling regime in a driven Fabry-Perot cavity. To do this, we will:
- Select a molecule.
- Calculate its absorption spectrum.
- Build the cavity.
- Simulate the cavity-molecule system.

## Calculate the Absorption Spectrum

To do this, we need to run two quick simulations. First, we use a short pulse (white light) without molecules. We then repeat the simulation with our molecule included.

### Empty-Box Case

To run the code, first create a working directory. We will call it `empty_box` and use the following command:

`mkdir empty_box`

Now create the `inp` file, which contains the main variables needed to run the simulation. We will use the `create.inp_file` function. To review the available variables, run:

```python
create.inp_file(help=True)
```

For now, we only need the following variables:

```python
dim         = 1
boundaries  = ["cpml"]
box_size    = [2000.0]
total_time  = 300.0
n_detectors = 2
print_time  = 0.05
dr          = 1.0

create.inp_file(mxll_dimensions=dim, mxll_boundaries=boundaries, mxll_box_size=box_size, 
                mxll_total_time=total_time, mxll_dr=dr, mxll_n_detectors=n_detectors, 
                mxll_dt_det_print=print_time)
```

We also need to include at least one source in the `sources.in` file. To do this, use:

```python
source    = "point"
pol       = "x"
field_amp = 1.0E-3
freq      = 7.0
tau       = 0.2
t0        = 5.0
t_init    = 0.0
t_end     = 10.0
r0        = [500.0, 0.0, 0.0]

create.sources_file(source_type=source, polarization=pol, field_amp=field_amp, 
                    frequency=freq, t0=t0, tau=tau, t_init=t_init, t_end=t_end, 
                    r0=r0, radius=0.0, phase=0.0)
```

We also need to specify the detector positions and the field to detect. In this case:

```python
detector_type = "point"
x_min         = -500.0

create.detectors_file(detector_type=detector_type, detected_field="Ex" , x_min=x_min)
create.detectors_file(detector_type=detector_type, detected_field="Hy" , x_min=x_min, file_exist=True)
```

Finally, move all input files to the `empty_box` working directory and run the executable there using the following commands:

`cp inp sources.in detectors.in empty_box/.`

`cd empty_box`

`export OMP_NUM_THREADS=4`

`/marado/installation/dir/build/bin/Marado.e`

The third command sets the number of OpenMP threads, which can speed up the calculation. You only need to do this once per session.

We can calculate the power transmitted at the detector position as:

$P(\omega) = -\mathrm{Re}\left\{ E_x^*(\omega) H_y(\omega) \right\}$

using the `process.plot_trans_spectrum_1D` function.

```python
process.plot_trans_spectrum_1D(directory="empty_box")
```

### Box with a Molecule

Now let's see what happens when we include the molecule.

First, create another working directory. Let's call it `box_with_1mol`.

`mkdir box_with_1mol`

Now create another `inp` file with the following variables:

```python
dim          = 1
boundaries   = ["cpml", "none", "none"]
box_size     = [2000.0, 0.0, 0.0]
total_time   = 300.0
n_detectors  = 2
n_q_groups   = 1
print_time   = 0.05
print_q_time = 0.05
mol_density  = 1.0E-3
dr           = 1.0
dt_q         = 0.1 # The time step of the quantum dynamics must at least double the time step of Mxll.

create.inp_file(mxll_dimensions=dim, mxll_boundaries=boundaries, mxll_box_size=box_size, 
                mxll_total_time=total_time, mxll_dr=dr, mxll_n_detectors=n_detectors, 
                mxll_dt_det_print=print_time, mxll_dt_q=dt_q, mxll_dt_q_print=print_q_time, 
                mxll_n_q_groups=n_q_groups, mxll_density_factor=mol_density)
```

Now create the `mol_group_xxxxxxx.in` file. Here, specify the number and positions of the molecules to simulate, the source of their input data, and whether to print their information:

```python
group_id    = 1
n_q_systems = 1
print_option = "print_all"
molecule_file_list = np.full(n_q_systems, 1) # All the molecules get the information from molecule_0000001.in
coordinates_list = np.array([0.0])

create.mol_group_file(file_number=group_id, number_of_q_systems=n_q_systems, print_options=print_option,
                      mol_file_list=molecule_file_list, coordinates=coordinates_list)
```

Then move all relevant files to the working directory:

`cp inp sources.in detectors.in molecule_0000001.in mol_group_0000001.in box_with_1mol/.`

You also need the `skf` files containing the parameters required for the calculation. These files are in the corresponding directories within `skf_files`. Copy them into `box_with_1mol`, then run the code from that directory.

If we plot the transmitted power again, we will not see much difference compared with the previous plot.

```python
process.plot_trans_spectrum_1D(directory="box_with_1mol")
```

To obtain the absorption spectrum, we need to calculate:

$P(\omega) = -\mathrm{Re}\left\{ (E_x^*(\omega)- E_{0,x}^*(\omega)) \, (H_y(\omega) - H_{0,y}(\omega)) \right\}$

The following function in `tools` performs this calculation automatically:

```python
process.plot_abs_spectrum_1D(only_pulse_dir="empty_box", sample_dir="box_with_1mol")
```

We can also plot properties of the molecule being simulated.

## Build the Cavity

Knowing the excitation wavelength, we can create a cavity that resonates with the molecule. To do this, we will simulate two mirrors using the Drude model:

$\frac{\partial^2 P_{x}(t)}{\partial t^2} + \gamma \frac{\partial P_x(t)}{\partial t} = \varepsilon_0 \Omega_p^2 E_x(t)$

We will use Drude parameters that are convenient for this tutorial, allowing access to frequencies that may not be reached with real metals. These parameters are $\Omega_p = 20.5 \, \rm eV$ and $\gamma = 0.18092 \, \rm eV$.

Now prepare a new working directory called `empty_cavity`:

`mkdir empty_cavity`

and create a new input file.

```python
dim         = 1
boundaries  = ["cpml"]
box_size    = [2000.0]
total_time  = 300.0
n_detectors = 2
n_media     = 2
print_time  = 0.05
dr          = 1.0

create.inp_file(mxll_dimensions=dim, mxll_boundaries=boundaries, mxll_box_size=box_size, 
                mxll_total_time=total_time, mxll_dr=dr, mxll_n_detectors=n_detectors, 
                mxll_n_media=n_media, mxll_dt_det_print=print_time)
```

We also need to create the `medium_xxxx.in` files, which specify the coordinates of the two mirrors. These coordinates must be consistent with the grid dimensions and spacing.

```python
Omega = 20.5 
gamma = 0.18092
dr = 1.0
mirror1_coordinates = np.arange(-60.0, -30.0+dr, dr)
mirror2_coordinates = np.arange(30.0, 60.0+dr, dr)

create.medium_file(file_number=1, medium_type="drude", omega=Omega, gamma=gamma, coordinates=mirror1_coordinates)
create.medium_file(file_number=2, medium_type="drude", omega=Omega, gamma=gamma, coordinates=mirror2_coordinates)
```

Move the input files to the working directory:

`cp inp sources.in detectors.in medium_0001.in medium_0002.in empty_cavity/.`

Now run the calculation again from the working directory, then plot the transmission spectrum.

```python
process.plot_trans_spectrum_1D(directory="empty_cavity")
```

The example cavity does not resonate at a frequency of interest, so adjust the separation between the mirrors. Keep in mind that the molecule is at the center of the simulation box.

Repeat the calculation until you obtain a cavity mode that resonates with the electronic transition of the chosen molecule.
